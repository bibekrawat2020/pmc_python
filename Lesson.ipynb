```python
from math import sqrt

n = int(input("Enter a number"))

if n < 2:
    print(n, "is not a prime number")
else:
    for i in range(2, int(sqrt(n)) + 1):
        if n % i == 0:
            print(n, "is not a prime number")
            break
    else:
        print(n, "is a prime number")
    
```

```python
def add_student():
    id = input("Enter ID: ")
    name = input("Enter Name: ")
    address = input("Enter Address: ")
    age = input("Enter Age: ")
    with open("students.csv", "a") as f:
        f.write(f"{id},{name},{address},{age}\n")
    print("Student added!")

def view_students():
    with open("students.csv", "r") as f:
        lines = f.readlines()
    if len(lines) == 0:
        print("No students found!")
    else:
        print("\nID  | Name  | Address | Age")
        print("-" * 30)
        for line in lines:
            id, name, address, age = line.strip().split(",")
            print(f"{id}  | {name}  | {address}  | {age}")

def search_student():
    search = input("Enter name to search: ")
    with open("students.csv", "r") as f:
        lines = f.readlines()
    found = False
    for line in lines:
        id, name, address, age = line.strip().split(",")
        if name.lower() == search.lower():
            print(f"\nID: {id}\nName: {name}\nAddress: {address}\nAge: {age}")
            found = True
            break
    if not found:
        print("Student not found!")

def menu():
    while True:
        print("\n===== Student Record System =====")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search Student")
        print("4. Exit")
        choice = input("Enter choice: ")
        if choice == "1":
            add_student()
        elif choice == "2":
            view_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid choice!")

menu()
```

```python
import os

def clear():
    os.system('cls' if os.name == 'nt' else 'clear')

def add_student():
    while True:
        id = input("Enter ID: ")
        name = input("Enter Name: ")
        address = input("Enter Address: ")
        age = input("Enter Age: ")
        with open("students.csv", "a") as f:
            f.write(f"{id},{name},{address},{age}\n")
        print("Student added!")
        ask = input("Do you want to add another student? (y/n): ")
        if ask.lower() != "y":
            break

def view_students():
    with open("students.csv", "r") as f:
        lines = f.readlines()
    if len(lines) == 0:
        print("No students found!")
    else:
        print("\nID  | Name  | Address | Age")
        print("-" * 30)
        for line in lines:
            id, name, address, age = line.strip().split(",")
            print(f"{id}  | {name}  | {address}  | {age}")

def search_student():
    search = input("Enter name to search: ")
    with open("students.csv", "r") as f:
        lines = f.readlines()
    found = False
    for line in lines:
        id, name, address, age = line.strip().split(",")
        if name.lower() == search.lower():
            print(f"\nID: {id}\nName: {name}\nAddress: {address}\nAge: {age}")
            found = True
            break
    if not found:
        print("Student not found!")

def search_by_id():
    while True:
        search_id = input("Enter ID to search: ")
        with open("students.csv", "r") as f:
            lines = f.readlines()
        found = False
        for line in lines:
            id, name, address, age = line.strip().split(",")
            if id == search_id:
                print(f"\nID: {id}\nName: {name}\nAddress: {address}\nAge: {age}")
                found = True
                break
        if not found:
            print("Student not found!")
        ask = input("Do you want to search by ID again? (y/n): ")
        if ask.lower() != "y":
            break

def delete_student():
    while True:
        search_id = input("Enter ID to delete: ")
        with open("students.csv", "r") as f:
            lines = f.readlines()
        found = False
        remaining = []
        for line in lines:
            id, name, address, age = line.strip().split(",")
            if id == search_id:
                found = True
            else:
                remaining.append(line)
        if found:
            with open("students.csv", "w") as f:
                f.writelines(remaining)
            print("Student deleted!")
        else:
            print("Student not found!")
        ask = input("Do you want to delete another student? (y/n): ")
        if ask.lower() != "y":
            break


def update_student():
    while True:
        search_id = input("Enter ID to update: ")
        with open("students.csv", "r") as f:
            lines = f.readlines()
        found = False
        updated = []
        for line in lines:
            id, name, address, age = line.strip().split(",")
            if id == search_id:
                print(f"Current: {id},{name},{address},{age}")
                name = input("New name: ")
                address = input("New address: ")
                age = input("New age: ")
                updated.append(f"{id},{name},{address},{age}\n")
                found = True
            else:
                updated.append(line)
        if found:
            with open("students.csv", "w") as f:
                f.writelines(updated)
            print("Student updated!")
        else:
            print("Student not found!")
        ask = input("Do you want to update another student? (y/n): ")
        if ask.lower() != "y":
            break

def menu():
    while True:
        clear()
        print("\n===== Student Record System =====")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search by Name")
        print("4. Search by ID")
        print("5. Update Student")
        print("6. Delete Student")
        print("7. Exit")
        choice = input("Enter choice: ")
        if choice == "1":
            add_student()
        elif choice == "2":
            view_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            search_by_id()
        elif choice == "5":
            update_student()
        elif choice == "6":
            delete_student()
        elif choice == "7":
            print("Goodbye!")
            break
        else:
            print("Invalid choice!")

menu()
```
